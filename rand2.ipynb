```python
def play_game():
    import random
    rn = random.randint(1,50)
    count = 0

    while True:
        user_rn = int(input("Enter your number (1-50) "))
        count += 1
        if(user_rn > rn):
            print(f"{user_rn} is High")
        elif(user_rn < rn):
            print(f"{user_rn} is Low")
        else:
            print(f"{user_rn} is correct, you guessed in {count} attempts")
            break                

play_game()
```
