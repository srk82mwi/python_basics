✅**Type cast to int or float at the time of input()**

```python
a = int(input("Enter value of a: "))
b = float(input("Enter value of b: "))

print(f"You have entered value of a: {a} {type(a)}")
print(f"You have entered value of b: {b} {type(b)}")

print(f"{a+b}")
```

<font color="yellow">**PYTHON OPERATORS**</font> <br>
✅ Operators are the symbols that perform some actions on the values <br>
✅<font color="red"> **Arithmetic Operators**</font>

|Op|Means      |Example|Result|
|--|-----------|-------|------|
|+ |add        |7+2    |9     |
|- |subtract   |7-2    |5     |
|* |multiply   |7*2    |14    |
|/ |divide     |7/2    |3.5   |
|//|floor division|7//2    |3     |
|% |modulus(remainder)        |7%2    |1     |
|** |power        |7**2    |49     |

```python
a = 7
b = 2
sum = a+b
sub = a-b
mul = a*b
div = a/b
fdiv = a//b
mo =a%b
po = a**b

print(f"{a} + {b} = {sum}")
print(f"{a} - {b} = {sub}")
print(f"{a} * {b} = {mul}")
print(f"{a} / {b} = {div}")
print(f"{a} // {b} = {fdiv}")
print(f"{a} % {b} = {mo}")
print(f"{a} ** {b} = {po}")
```

<font color="yellow">**COMPOUND ASSIGNMENT OPERATOR** </font> <br>
✅Combines assignment operator with arithmetic operators <br>
|Operator|Example|Output|
|--------|-------|------|
|+= |x+=3|x=x+3|
|-= |x-=3|x=x-3|
|*=|x*=3|x=x*3|
|/=|x/=3|x=x/3|   

```python
score = 50
score += 5
print(f"score={score}")
score -= 35
print(f"score={score}")
score *= 4
print(f"score={score}")
score /= 2
print(f"score={int(score)}")
```

<font color="yellow"> **COMPARISON OPERATOR** </font> ⚖️ <br>
✅asks yes/no questions returns TRUE/FALSE <br>

|Op|Means|
|--|-----|
|==|equal to|
|!=|not equal to|
|<,<=|less than, less than equal to|
|>,>=|greater than, greater than equal to|




```python
print(f"10==10 -> {10==10}")
print(f"10!=10 -> {10!=10}")
print(f"10>=1 -> {10>=1}")
print(f"10<=7 -> {10<=7}")
```

<font color="yellow"> **LOGICAL OPERATORS (and, or, not)** </font>
<br>

|Operation|Means|Operation|
|---------|-----|---------|
|and|return true if both conditions are true|10>3 and 5<7 -> True|
|or|returns true if any condition is true|10>3 or 5>7 -> True|
|not|flips true/false|not(10>3) -> False|

```python
print(f"10>3 and 5<7 -> {10>3 and 5<7}")
print(f"10>3 or 5>7 -> {10>3 or 5>7}")
print(f"not(10>3) or 5>7 -> {not(10>3) or 5>7}")
```

<font color="yellow"> **MEMBERSHIP OPERATORS (in, not in)** </font> <br>
✅ used to search something inside a string

```python
name = "shafique ur rehman"
print(f"Does 'e' exist in '{name}' -> {'e' in name}")
print(f"Does 'c' not exist in '{name}' -> {'c' not in name}")
print(f"Does 'a' exist in '{name}' -> {'a' in name}")
```

```python
fruits =['orange',"mango",'apple']
fav_fruit = 'grapes'
print(f"Does {fav_fruit} not exist in {fruits} -> {fav_fruit not in fruits}")
print(f"Does 'banana' not exist in {fruits} -> {'banana' not in fruits}")
print(f"Does 'm' exist in {fruits} -> {'m' in fruits}")
print(f"Does 'mango' exist in {fruits} -> {'mango' in fruits}")
```
