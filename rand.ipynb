```python
import random

rn = random.randint(1,20)
count = 0
user_rn = int(input("Guess the number(1-20): "))
count = 1
while(user_rn != rn):
    if(user_rn > rn):
        print(f"{user_rn} is High")
    elif(user_rn < rn):
        print(f"{user_rn} is Low")
    count += 1
    user_rn = int(input("Guess the number(1-20) "))

print(f"Secret number is {user_rn}, you guessed in {count} attempts")
```
