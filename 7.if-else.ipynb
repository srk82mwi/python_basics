Logical Operators
- and --> return true if both conditions are true
- or --> returns true if any one condition is true
- not --> flips True/False

**<mark>AND</mark> Operator**

```python
age = int(input("Enter your age: "))
has_ticket = True
print(f"User's age is {age}")
if age>=18 and has_ticket:
    print("You are allowed, Welcome to the show")
else:
    print("Sorry! you are not allowed") 
```

**<mark>OR</mark> Operator**

```python
day = "sunday"

if day.lower() == 'saturday' or day.lower() =='sunday':
    print("it's weekend")
else:
    print("it's a weekday.")
```

**<mark>NOT</mark> Operator**

```python
is_raining = True

if not is_raining:
    print("You can go outside")
else:
    print("You can't go outside")
```

**Check Membership inside condition
using <mark>in</mark> Operator**

```python
fruits = ['apple','banana','mango']
print('mango' in fruits)
```

**Search item in a list using <mark>in</mark> operator**

```python
fruits = ['apple','banana','mango'];
user_fruit = input("Enter your fruit: ")

if user_fruit in fruits:
    print(f"Yes, we have {user_fruit}")
else:
    print(f"Sorry, we don't have {user_fruit}")
```

**<mark>Search</mark> text in a string**

```python
name = "shafique ur rehman"
name_2 = "sha"
print(name==name_2)  #checks whether both strings are equal
print(name_2 in name) #finds a string in another string
```
