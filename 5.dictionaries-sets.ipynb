# **Class 5 - Python Dictionaries and Sets**

**key:pair values**

```python
students = {"rollno":100,"name":"ali","group":"FA"}
print(students["name"])
print(students["rollno"])
#print(students["address"])
print(students.get("group"))
```

**Check data type**

```python
print(type(students))
```

**Add new key pair value in dictionary**

```python
students["address"]="isakhel"
print(students.get("address"))
```

```python
print(students)
```

**Update key pair value in dictionary**

```python
students['group']='ICS'
students
```

**Deleting a dictionary**

```python
del students
students
```

**Deleting a key pair value**

```python
students = {"rollno":100,"name":"ali","group":"FA","address":"isakhel"}
students
```

```python
del students['address']
```

```python
students
```

**Looping a dictionary**

```python
# only returns keys not values
for n in students:
    print(n)
```

```python
students.items()
```

```python
students.keys()
```

```python
for n in students.keys():
    print(n)
```

```python
students.values()
```

```python
for n in students.values():
    print(n)
```

```python
for key,val in students.items():
    print(f"{key} = {val}")
```
